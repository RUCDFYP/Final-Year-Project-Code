# The code creates the skelleton of folders and subfolders for the project


```python
# Step 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Step 2: Install required packages
!pip install pandas openpyxl --quiet

# Step 3: Upload the Excel file
from google.colab import files
uploaded = files.upload()

# Step 4: Read the Excel file
import pandas as pd
import os

# Replace with your uploaded filename
excel_filename = list(uploaded.keys())[0]
df = pd.read_excel(excel_filename)

# Step 5: Define root folder path
root_folder = '/content/drive/MyDrive/AA_Project_Data'

# Step 6: Create folder structure
def create_folder_structure(df):
    if not os.path.exists(root_folder):
        os.makedirs(root_folder)
        print(f"Created root folder: {root_folder}")
    else:
        print(f"Root folder already exists: {root_folder}")

    for _, row in df.iterrows():
        company = str(row['Pharma Company']).strip()
        ceo = str(row['CEO']).strip()

        company_path = os.path.join(root_folder, company)
        ceo_path = os.path.join(company_path, ceo)

        # Create company folder if it doesn't exist
        if not os.path.exists(company_path):
            os.makedirs(company_path)
            print(f"Created company folder: {company_path}")

        # Create CEO folder if it doesn't exist
        if not os.path.exists(ceo_path):
            os.makedirs(ceo_path)
            print(f"Created CEO folder: {ceo_path}")

        # Create subfolders inside CEO folder
        subfolders = ['Raw_Data', 'Preprocessed_Data', 'LIWC_Output', 'Big_Five_Scores']
        for subfolder in subfolders:
            subfolder_path = os.path.join(ceo_path, subfolder)
            if not os.path.exists(subfolder_path):
                os.makedirs(subfolder_path)
                print(f"Created subfolder: {subfolder_path}")

# Step 7: Run the folder creation
create_folder_structure(df)
print("✅ Folder structure creation complete.")
```
